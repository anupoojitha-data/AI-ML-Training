```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## 1. Load the dataset

```python
df = pd.read_csv('adult.csv')
df.head()
```

```python
df.info()
df.shape
```

```python
df.describe()
```

## 2. Data Cleaning

```python
# Check missing values
df.isnull().sum()
```

```python
# In this dataset, some missing values are written as '?'
df = df.replace('?', pd.NA)
df.isnull().sum()
```

```python
# Fill missing categorical values with 'Unknown'
df = df.fillna('Unknown')
df.isnull().sum()
```

```python
# Check and remove duplicates
print('Duplicates:', df.duplicated().sum())
df = df.drop_duplicates()
print('Shape after cleaning:', df.shape)
```

```python
# Check obvious outliers in age and hours worked
print(df[['age', 'hours-per-week']].describe())

# Remove only impossible values; keep valid extreme values
df = df[(df['age'] >= 17) & (df['age'] <= 100)]
df = df[(df['hours-per-week'] >= 1) & (df['hours-per-week'] <= 99)]
df.shape
```

## 3. Exploratory Data Analysis (EDA)

Four simple charts: histogram, boxplot, count plot and correlation heatmap.


```python
# Chart 1: Age distribution
sns.histplot(df['age'], bins=20)
plt.title('Age Distribution')
plt.show()
```

```python
# Chart 2: Hours worked by income
sns.boxplot(x='income', y='hours-per-week', data=df)
plt.title('Working Hours vs Income')
plt.show()
```

```python
# Chart 3: Income class count
sns.countplot(x='income', data=df)
plt.title('Income Distribution')
plt.show()
```

```python
# Chart 4: Correlation heatmap
plt.figure(figsize=(8, 5))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

## 4. Feature Engineering

```python
# New feature 1: Total capital = gain - loss
df['net_capital'] = df['capital-gain'] - df['capital-loss']

# New feature 2: Working hours category
df['work_type'] = df['hours-per-week'].apply(
    lambda x: 'Full-time' if x >= 40 else 'Part-time'
)

df[['capital-gain', 'capital-loss', 'net_capital', 'hours-per-week', 'work_type']].head()
```

## 5. Five Key Insights

```python
# These 5 findings use actual values from the dataset when you run the notebook.
print('1. Most common income category:', df['income'].mode()[0])
print('2. Average age:', round(df['age'].mean(), 1))
print('3. Average weekly working hours:', round(df['hours-per-week'].mean(), 1))
print('4. Most common education level:', df['education'].mode()[0])
print('5. Percentage working full-time:',
      round((df['work_type'] == 'Full-time').mean() * 100, 1), '%')
```

Run the cell above and use its results to complete these findings:
